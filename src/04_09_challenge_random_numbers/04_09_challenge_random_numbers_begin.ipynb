```python
from qiskit import *
from qiskit.visualization import plot_histogram, plot_bloch_multivector
from math import pi
%matplotlib inline

#Project intent: configure states of qubits
#to have the following distribution:
#|001> = 0.375, |011> = 0.375
#|101> = 0.125, |111> = 0.125
```

```python
circuit = QuantumCircuit(3)

#Trial 1
circuit.rx(pi, 0) #Makes 1st qubit always in 1 position
circuit.h(1) #Makes 2nd qubit equally 1 & 0
circuit.rx(pi/3, 2) # Makes 3rd qubit 75% 0

#YOU'RE A F*CK*N GENIUS BRO WELL DONE
```

```python
circuit.remove_final_measurements
from qiskit.quantum_info import Statevector
statevector = Statevector(circuit)
plot_bloch_multivector(statevector)
```

```python
circuit.measure_all()
circuit.draw(output='mpl')
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
from qiskit.providers.basic_provider import BasicProvider
simulator = BasicProvider().get_backend('basic_simulator')
compiled_circuit = transpile(circuit, simulator)
result = simulator.run(compiled_circuit, shots=1000000).result()
plot_histogram(result.get_counts())
```
